```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Titanic dataset
df = pd.read_csv(r"C:\Users\Admin\Downloads\titanic.csv")

print("First 5 rows:")
print(df.head())
print("\nDataset Shape:", df.shape)

sns.set_theme(style="whitegrid")

# 1. LINE PLOT: Average Fare by Passenger Class
fare_by_class = df.groupby("Pclass")["Fare"].mean()

plt.figure(figsize=(8, 5))
plt.plot(fare_by_class.index, fare_by_class.values,
         marker="o", color="blue", linewidth=2)
plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.xticks([1, 2, 3])
plt.tight_layout()
plt.show()

# 2. BAR PLOT: Survival Count
survivors = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
sns.barplot(x=survivors.index, y=survivors.values,
            color="orange")
plt.title("Titanic Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()

# 3. SCATTER PLOT: Age vs Fare
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Age", y="Fare",
                hue="Survived", alpha=0.7)
plt.title("Passenger Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()

# 4. HEATMAP: Correlation Matrix
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(10, 7))
sns.heatmap(numeric_df.corr(), annot=True,
            cmap="coolwarm", fmt=".2f")
plt.title("Titanic Dataset Correlation Heatmap")
plt.tight_layout()
plt.show()

print("All four plots generated successfully!")
```
