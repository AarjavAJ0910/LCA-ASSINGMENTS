```python
# Create a dictionary with 5 elements
my_dict = {
    "name": "Aarjav",
    "rollno": 34,
    "branch": "AI & DS",
    "age": 18,
    "marks": 85
}

print("Original Dictionary:")
print(my_dict)

# Add student information
my_dict["city"] = "Indore"

print("\nAfter adding city:")
print(my_dict)

# Delete student information
del my_dict["age"]

print("\nAfter deleting age:")
print(my_dict)

# Display student information
print("\nStudent Information:")
for key, value in my_dict.items():
    print(key, ":", value)
```

```python
list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]

my_dict = dict(zip(list1, list2))

print("Dictionary:")
print(my_dict)
```

```python
my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}

print("Dictionary:")
print(my_dict)

print("\nSorted elements:")
print(sorted(my_dict))
```

```python
mydict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

values = mydict.values()

mean = sum(values) / len(values)

print("Mean of all values:", mean)
```

```python
my_dict = {
    "name": ["Yash", "Neel", "Dev"],
    "rollno": [11, 12, 13],
    "marks": [78, 56, 98]
}

# a) Display name as Dev
print("Name:", my_dict["name"][2])

# b) Display 12 roll no
print("Roll No:", my_dict["rollno"][1])

# c) Display greatest marks
print("Greatest Marks:", max(my_dict["marks"]))
```

```python
string = input("Enter a string: ")

frequency = {}

for letter in string:
    if letter in frequency:
        frequency[letter] += 1
    else:
        frequency[letter] = 1

print("Frequency of letters:")

for key, value in frequency.items():
    print(key, ":", value)
```

```python
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)

print("Tuple of Quotient and Remainder:", result)
```
